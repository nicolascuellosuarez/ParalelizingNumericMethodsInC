## **Una Vista Sobre la Dinámica Térmica.**

Realicemos un ejercicio simple: Imagina una **barra de metal** cuya longitud es, digamos, del tamaño de una *regla escolar*. Con una mano, la sostienes por un **extremo**, con la otra, aproximas un encendedor otro extremo para calentarlo. No lo apoyas, el fuego no toca la barrra, solo lo acercas; al poco tiempo ves que el metal se calienta **justo donde está la llama**. Teniendo en cuenta esta situación, una pregunta válida que puede surgir es:

- **Si me quedo en esta posición, ¿cuánto tardará en sentirse el calor en la mano que uso para sostener el otro extremo?**

Interpolando lo anterior a cualquier otra situación que se relacione con el **fuego**, por simple **empiria**, sabemos que la propagación del fuego no es **instantánea** ni mucho menos, el calor viaja **de forma lenta** pero constante, no salta ni vuela, más bien **se difunde**. Este es un conocimiento básico y, como lo dije antes, empírico, pero, si somos curiosos y nos vamos más allá de lo que observamos, preguntas que pueden surgir son:

- **¿Podemos estimar, de alguna manera, el tiempo en el que el extremo frío se calentará?**

## **La Matemática que Necesitaremos.**

Antes de lanzarnos de lleno a **modelar cómo el calor se mueve**, tenemos que asegurarnos de tener las herramientas adecuadas para la **elaboración** y **comprensión del modelo**. Al igual que no se puede construir una casa sin saber utilizar un martillo, no se puede **modelar la realidad** sin tener claros ciertos conceptos clave.

Si queremos describir el viaje del **calor** sobre un objeto como una **barra metálica**, debemos tener en cuenta que el **fenómeno** de transmisión depende de dos cosas al mismo tiempo: la **posición** a lo largo de la barra y el **instante** en el que medimos la temperatura. Eso significa que nuestra **función** de temperatura **no puede** ser definida solo por una variable, en este caso es **indispensable** que se tengan en cuenta esos dos factores. Aquí necesitamos algo más elaborado: una función que recibe dos números **(dónde y cuándo)** y devuelve un número **(la temperatura)**. Existe una herramienta matemática que nos permite modelar este comportamiento, y son las **Funciones de Dos Variables**, una de las bases más fuertes del **Cálculo Multivariable**; a ese tipo de funciones, las podemos escribir de la siguiente manera:

$$f(x,t)$$

Donde la letra $f$ corresponde a la **función de temperatura**, la $x$ corresponde a la **posición de un punto de la barra**, y la $t$ se refiere a el **tiempo transcurrido**. En papel es simple, ¿no? Es simplemente agregarle una pequeña cosita más a las funciones simples que ya conocemos, sin embargo, detrás de esa simpleza, se esconde una complejidad **digna de admiración**: ahora tenemos herramientas para contestar cuestiones que involucren *el cambio* de esas dos variables, como por ejemplo, **¿cómo cambia la temperatura si dejo pasar un instante sin moverme de mi lugar?**, **¿Cómo cambia la temperatura si me muevo a cierta posición, sin haber pasado un instante?**, aquí es donde entra otra herramienta poderosísima: *la derivada parcial*.

Recordemos, brevemente, lo que significa el concepto de la **derivada**, evitando las formalidades, como el uso de límites y otras herramientas **formales** explicativas. Recordemos el concepto de la derivada de forma analítica. Si tenemos una función **univariable**, de toda la vida, definida como $f(x)$, si hallamos la **derivada** $f'(x)$ de esa función, vemos una información verdaderamente diciente sobre la función: **¿qué tan rápido está cambiando el valor de la función cuando movemos un poquito la $x$?** Si la derivada es grande y positiva, la función está subiendo bruscamente. Si es negativa, está bajando. Si es cero, está plana, al menos por un instante. Geométricamente, la derivada es la **pendiente** de la **recta tangente** a la curva en ese punto. Una **tangente** significa "que toca el objeto en un punto", como una regla que apoyamos suavemente sobre el **borde de una montaña**.

Eso es claro, **sencillo** y básico, pues son los fundamentos del **cálculo diferencial**, sin embargo, si tenemos en cuenta lo visto anteriormente, es completamente natural preguntarse: **ya, pero, ¿y si tenemos dos variables en una funcion?, ¿cómo se deriva en ese caso?** Ahí, la cosa se vuelve más interesante, pues no tenemos una única pendiente, tenemos dos:

1. Una que nos **indica** el cambio en la función si nos movemos en la **dirección de $x$** (dejando al tiempo $t$ fijo).
2. Otra que nos indica el cambio en la función si nos movemos en la **dirección de $t$** (dejando a la posición $x$ fija).

A eso, se le llaman **Derivadas Parciales**, una herramienta matemática que nos permite ver el cambio infinitamente instantáneo de una sola **variable** en una función multivariable, tratando como constantes a los otros elementos que definen a la función, si unimos la **derivada parcial** de cada una de las variables de una función en un **vector** $v$, lograremos un **vector** (flecha), que nos indica la dirección en la que la función más está creciendo; podemos definir a este tipo de derivadas de la siguiente manera **(formalmente)**:

$$\frac{\partial u}{\partial x}, \frac{\partial u}{\partial t}$$

El símbolo $\partial$ nos recuerda que **estamos derivando respecto a una variable** mientras tratamos a la otra como si fuera **constante**. Como si congeláramos el tiempo y solo miráramos cómo varía la **temperatura** a lo largo de la barra, o como si nos fijáramos en un punto fijo y observáramos cómo se calienta o se enfría con el paso de los segundos. Por ejemplo, definamos a la siguiente función bivariable, cuya gráfica debe representarse en un espacio de 3 dimensiones ($\mathbb{R}³$), tal que:

$$f(x, y) = 2x + y²$$

Como vemos, es una función que recibe **dos entradas** para dar una respuesta, pues debemos ingresar un valor para la $x$ y otro para la $y$. Entonces, en base a lo que hemos desarrollado anteriormente, **¿cómo derivamos entonces a esta función, si tiene dos variables?**, simple, utilizamos el conocimiento adquirido, y derivamos parcialmente. Primero, derivemos parcialmente la variable $x$, esto se hace, utilizando a cualquier variable $y$ como constante:

$$\frac{\partial}{\partial x} (2x + y²) = 2$$

Bien, tenemos una derivada parcial, ahora **apliquemos lo mismo**, pero para la variable $y$, derivando solo los términos que contengan a esa variable, y utilizando a los términos en $x$ como una constante:

$$\frac{\partial}{\partial y} (2x + y²) = 2y$$

Entonces, el **gradiente**, o el **vector** de dirección de la tasa de mayor **crecimiento** de la función en cualquier punto, denotado con el símbolo $\nabla$, corresponde a un vector en el que almacenamos las dos derivadas parciales calculadas previamente, tal que:

$$\nabla f(x, y) = (2, 2y)$$

Ahora, teniendo en cuenta estas ideas, podemos entrar a **pensar** en muchas más cosas, pues ya tenemos **ciertas herramientas** para poder responder preguntas importantes sobre el paso del calor. Como vimos, ya tenemos una función **bi-variable** que nos permite conocer la temperatura en base a la **posición** ($x$) en la barra y el **tiempo** ($t$), además, logramos visualizar una manera en la que podemos hallar el cambio que existe en la propagación de ese calor, utilizando el concepto de **derivadas parciales**. Sin embargo, todavía no estamos del todo listos para el trabajo, pues nos falta pensar en algo de vital importancia, que no estamos teniendo en cuenta; a ver, si la **temperatura** en un punto de la barra puede cambiar a medida que pasa el **tiempo**, o porque nos movemos hacia otro lugar (como definimos en nuestra **función de temperatura**), **¿cómo podemos relacionar esas dos formas de cambio?**, **¿Qué ecuación modela esta relación?**

La respuesta a esto es, **en esencia**, un concepto **matemático** de los más fascinantes e importantes en la historia, pues es uno de los motores por los cuales podemos **abstraer la realidad** y entenderla de forma matemática: **Las Ecuaciones Diferenciales Parciales**.

Sabemos que, de forma básica, una **ecuación normal**, de toda la vida, es una expresión en la que tenemos dos **miembros** (**expresiones**) separadas por un símbolo $=$, que contiene una o más incógnitas o (valores desconocidos); las **ecuaciones** son una forma básica de hallar respuestas a incógnitas o inquietudes con respecto a una expresión, y son herramientas que nos permiten hallar las **raíces** o **interceptos** de una parábola o cualquier polinomio, hallar múltiples incógnitas si hablamos de un **sistema de ecuaciones**, etc. Básicamente, estamos frente a una de las herramientas matemáticas más útiles en cualquier campo. Sin embargo, extraigamos la esencia de la ecuación y pensémosla en este plano: necesitamos hallar una relación que nos permita conocer cómo puede cambiar la **temperatura** con respecto al **tiempo** o la **posición**, es decir, aquí la incógnita no es una sola variable, es el comportamiento de una función entera.

Entonces pensemos, ¿qué tal si, en vez de que una **ecuación** resuelva el valor de una **incógnita única**, resuelve más bien la forma en la que una **función se comporta**?, **¿y si la incógnita no es solo una variable, si no que es una función entera?**. Esto, básicamente, es la intuición de las **ecuaciones diferenciales**, una ecuación en la que relacionamos una función desconocida, sus variables independientes y sus **derivadas**, no queremos hallar $x$, queremos hallar la forma de $f(x, t)$ tal que al derivarla de cierta manera, logremos una expresión en específico. Ahora, existen diferentes tipos de **ecuaciones diferenciales**, si la ecuación depende de **una sola variable** (como solo la posición o el tiempo), hablamos de **Ecuaciones Diferenciales Ordinarias (EDO)**, y la incógnita es $f(y)$ o $f(t)$, sus derivadas totales serían $f'(y)$ o $f''(t)$.

Pero cuando la función a encontrar se encuentra definida por **2 o más variables**, como es nuestro caso con la ecuación del calor, estamos hablando de **Ecuaciones Diferenciales Parciales (EDP)**; la ecuación aquí tratada es una de las más famosas y más **fáciles** de escribir, sin embargo, no siempre es una de las más faciles de resolver.
