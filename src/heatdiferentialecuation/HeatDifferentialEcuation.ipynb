## **Una Vista Sobre la Dinámica Térmica.**

Realicemos un ejercicio simple: Imagina una **barra de metal** cuya longitud es, digamos, del tamaño de una *regla escolar*. Con una mano, la sostienes por un **extremo**, con la otra, aproximas un encendedor otro extremo para calentarlo. No lo apoyas, el fuego no toca la barrra, solo lo acercas; al poco tiempo ves que el metal se calienta **justo donde está la llama**. Teniendo en cuenta esta situación, una pregunta válida que puede surgir es:

- **Si me quedo en esta posición, ¿cuánto tardará en sentirse el calor en la mano que uso para sostener el otro extremo?**

Interpolando lo anterior a cualquier otra situación que se relacione con el **fuego**, por simple **empiria**, sabemos que la propagación del fuego no es **instantánea** ni mucho menos, el calor viaja **de forma lenta** pero constante, no salta ni vuela, más bien **se difunde**. Este es un conocimiento básico y, como lo dije antes, empírico, pero, si somos curiosos y nos vamos más allá de lo que observamos, preguntas que pueden surgir son:

- **¿Podemos estimar, de alguna manera, el tiempo en el que el extremo frío se calentará?**
- **¿Existe alguna manera de conocer la temperatura de un punto cualquiera de la barra?**
- **¿Qué herramientas podemos utilizar en este caso?**

Este tipo de **curiosidades**, que en primera instancia uno pensaría que son tontas e inútiles, son las cuestiones que han motivado el avance **científico**, son preguntas completamente habituales y casi de obligada ejecución en un **contexto de ciencia**; son tan habituales que alguien, hace mucho tiempo atrás, en el siglo **XIX**, logró hallar una respuesta concreta y demostrable, el nombre es **Jean Fourier**, un matemático francés cuya mente fue una de las más brillantes en su época.

**Fourier** no solo encontró una resolución a estas dudas instrumentalizando un lenguaje universal, las **matemáticas** (creando una ecuación que modela el viaje del calor), si no que logró una forma de pensar que terminó siendo mucho más grande que este simple problema, pues es **transversal** a muchas otras cuestiones.

En este notebook realizaremos una **explicación profunda** y detallada de cada uno de los **elementos matemáticos** que conlleva la resolución de este problema, así como la consideración intuitiva del modelo que lo abstrae, logrando una solución que se ejecuta a partir de una herramienta de poder enorme: **la programación**, explicando cada algoritmo, método y herramienta utilizada, tanto para su **solución computacional** como para la optimización de esa respuesta. Básicamente, entenderemos de forma **física** e **intuitiva** el significado de que el calor se *funda*, para traducirlo a un **modelo matemático** cuyas soluciones pueden hallarse haciendo uso de la **fuerza computacional**.
