## **Las Bases del Cálculo de Áreas.**

Si estamos en un contexto universitario, el "ABC" del ingeniero corresponde a el conocimiento en los inicios paradigmáticos del cálculo, es decir, todo se construye a partir de principios tan claves como lo son los procesos de **derivación** y **antiderivación**. Formalmente, conocemos a la derivada como el cambio infinitamente inmediato que existe en una función univariable, vemos que la definición formal de la derivada corresponde a:

$$f'(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$$


Es decir, un límite de $h$ a $0$ en el que $h$ corresponde al cambio **(en $x$)** de la función; este proceso nos permite hallar la **recta tangente** a un punto específico en una función, pues la diferencia se hace infinitamente pequeña. A partir de esto, nacen múltiples métodos de derivación para hallar las derivadas de cualquier función (desde funciónes *polinómicas* a *trigonométricas*, *inversas trigonométricas*, etc.); Esto nos hace notar algo, y es que, al derivar una función, estas aplicando ciertas reglas que idealmente, pueden deshacerse, es decir, al derivar, **estás transformando una función a partir de reglas**. 

Si hacemos un ejercicio mental breve, podemos concluir que, si podemos derivar una función, podemos ***antiderivarla*** luego. Así pues, tenemos otra herramienta clave, la **antiderivada (o primitiva de una función)** como operación inversa a la **derivada**, que se define como el proceso por el que se halla una función primitiva **$F(x)$**, que al derivarla, logramos nuevamente la función original **$f(x)$** tal que:

$$F'(x) = f(x)$$

Y si **$F(x)$** es una **antiderivada** particular, podemos expresar la operación de **antiderivada** mediante la **integral indefinida** que representa una **familia de curvas**:

$$\int f(x) \ dx = F(x) + C$$

Así pues, construimos herramientas poderosas para el **análisis**, sin embargo, todo esto está basado en el uso del álgebra para la transformación de funciones definidas; no logramos una pista para lograr valores numéricos que representen áreas en intervalos específicos, hasta ahora, es mero uso del álgebra simbólica para operar funciones tanto **derivándolas** como **antiderivándolas**.

Para responder a esta necesidad, nace una herramienta mucho más poderosa aún, la **integral definida**, que nos permite, básicamente, hallar el **área** que existe debajo de la **curva o recta** de una función **$f(x)$**, en un **intervalo específico**. 

**¿Cómo se calcula esta área?**: muy simple, a partir de las herramientas construidas anteriormente: El método clásico y de perogrullo que habitualmente es enseñado, es el de, hallar la **integral indefinida** de la función cuya área se quiere encontrar, y posterior a eso, **evaluar** en unos límites de integración definidos (**$a$** y **$b$**) que corresponden a los extremos del intervalo a analizar, tal que:



$$\int_{b}^{a} f(x) \ dx$$

Esto funciona porque es el equivalente a una herramienta matemática que estabamos pasando por alto y es crucial para el cálculo de áreas: **La Sumatoria de Riemann**, una **sumatoria básica** en la cual dividimos el intervalo a calcular en un número de **rectángulos** específico, y sumamos el área (base X altura) de cada uno de ellos, siendo la **altura**, el valor de la función en ese punto y la base, la base del rectángulo creado; entre más divisiones (rectángulos) halla, vamos a lograr un resultado más preciso y apegado a la realidad. Así pues, pasamos de cálculos más complicados algebráicamente (como los procesos de **derivación** y **antiderivación**), a algo mucho más simple, como lo es la sumatoria de áreas más fáciles de calcular. Podemos definir, entonces, a la **integral definida** como la **Suma de Riemann** cuando la base de esos **rectangulos** tiende a $0$, es decir, cuando la cantidad de rectángulos tiende a infinito:

$$\int_{b}^{a} f(x) \ dx = \lim_{n \to \infty} \sum_{i = 1}^n f(x_i^*) \Delta x$$

Esto es contrastado y funcional, un recurso matemático resolutivo y útil para multiples necesidades: 

- Hallar **áreas** contenidas entre una o diferentes curvas.
- Hallar **volúmenes** de sólidos de revolución.
- Hallar el **trabajo** necesario para elevar fluídos a cierta altura.
- Hallar la **fuerza hidrostática** que ejerce un fluído sobre x figura.

Así pues, definimos a la integral definida como una sumatoria de áreas infinitesimalmente pequeñas. Sin embargo, esto nos hace crear nuevas preguntas:

1. **¿Todas las áreas pueden hallarse de esa forma?**
2. **¿Todas las funciones tienen una antiderivada primitiva que puede expresarse?**
3. **¿Es esta la forma más eficiente de hallar áreas?**
4. **¿Se pueden computar estos métodos?, ¿Son computacionalmente óptimos?**

## **La Integración Numérica: Una Optimización Pura.**

Lo visto anteriormente es una generalización matemática acertada y descrita por los genios más grandes de la historia; es funcional para todos y cada uno de los casos, todas las áreas pueden hallarse utilizando lo anterior. **¿O no?**, probémoslo, utilicemos una función arbitraria cualquiera para hallar su antiderivada y evaluar en límites. Probemos con una función simple:

$$f(x) = \cos(x²)$$ $$\int \cos(x²) \ dx$$

**¿Podemos utilizar la integral indefinida?**, fácil, sustitución simple, **¿o no?**: no, pues al utilizar a $x²$ como $u$ y derivarla, la integral queda en dos variables, no podemos liberarnos. 

**¿Sustitución trigonométrica?**, no, pues no hay identidad que permita convertir a la integral en una forma que permita esta sustitución. 

**¿Integración por Partes?**, no, entramos en un bucle infinito.

Entonces, ¿cómo lo integramos?; ahí, acabamos de entrar a un factor importante, y es que existen funciones cuya **antiderivada existe**, pero no puede ser expresada con una fórmula finita conocida fácilmente. Es decir, sabemos que existe un área debajo de la curva de $\cos(x²)$, sin embargo, no existe una combinación básica de funciones cuya derivada sea $\cos(x²)$. Entramos al campo de las **Integrales NO elementales**.

Además, pensemos en algo, si queremos optimizar la resolución de un problema, lo más facil es programarlo, así podemos generalizar y conseguir respuestas de forma mucho más óptima. Entonces, ¿podemos computabilizar este problema?, según lo que hemos dicho, necesitaríamos poder programar antiderivadas. Eso existe, se llama **Álgebra Computacional**, sin embargo, preguntémonos, **¿es esta la forma más óptima de solucionar problemas de grandes volúmenes de datos?**, **¿qué tan fácil es programar álgebra simbólica?**.

Ambas respuestas nos dan un mal augurio, puesto que para programar **álgebra simbólica computacional**, se deben manipular caracteres, reglas lógicas y transformaciones que consumen tiempo de procesamiento y memoria; simplemente para antiderivar una función como:

$$f(x) = \sin²(x)\cos³(x)$$

Se requiere de múltiples transformaciones, identidades y sustituciones, lo que hace de estos algoritmos algo complejo e ineficiente de manipular, por todas las consideraciones que se deben tener. Por eso, herramientas como ***Mathematica*** o ***SymboLab*** son tan pesadas. Además, se deben implementar algoritmos sumamente complejos, extensos y poco ortodoxos. 

Es por eso, que existe una respuesta óptima, funcional, y muy buena ante estas dificultades: **el cálculo numérico**, un conjunto de algoritmos que utilizan operaciones aritméticas simples como la **suma** o **la multiplicación** para lograr resultados muy bien aproximados y convergentes para problemas matemáticos mucho más complejos. Se debe tener un análisis más simplista de los problemas para resolverlos, sin embargo, pueden ser computacionalmente mucho más óptimos y fáciles de resolver, pues se basan en operaciones básicas y más sencillas para un procesador.

Por ejemplo, la sumatoria de Riemann nos permitiría hallar el área debajo de la curva en un intervalo para **Integrales No Elementales**, pues es un problema fácil de computar, una sumatoria de las áreas de $n$ cantidad de rectángulos. Así pues, vemos que el problema anteriormente planteado se soluciona de una forma simple: **utilizando operaciones básicas para discretizar procesos complejos**, incluso, podemos implementar directivas de paralelización para una resolución de esa índole, sin embargo, debemos, nuevamente, preguntarnos lo siguiente:

- **¿Es esta resolución para el cálculo de áreas la más óptima en términos computacionales?...**

Hagamos un ejercicio para responder a la disyuntiva anterior, probemos calcular el área de $f(x) = e^{-x²}$ en el intervalo cerrado de $[-1, 1]$, tal que:

$$\int_{-1}^{1} e^{-x²}\ dx$$

Para ejecutar este cálculo, debemos primero pensar en el algoritmo que nos va a permitir realizarlo; en este caso, podemos plantearlo de una manera muy simple; si implementamos un código serial, utilizando C, para hallar el área debajo de la curva de la función dada en el intervalo solicitado. 

Lo hacemos de la siguiente manera:

```c
//%ldflags: -lm
// Flag utilizada para incluir -libm
#include <stdio.h>
#include <math.h>
#include <omp.h>

double function(double x) {
    return exp(-(x*x));
}

double calculate_integral_using_riemann_sum(double inferior_limit, double superior_limit, double rectangles) {
    double sum = 0;

    if (inferior_limit < superior_limit) {
        double distance_between_limits = superior_limit - inferior_limit;
        double split = distance_between_limits / rectangles;

        while (inferior_limit < superior_limit) {
            inferior_limit += split;
            sum += function(inferior_limit) * split;
        }

        return sum;
    } else if (inferior_limit == superior_limit) {
        return 0;
    } else {
        double distance_between_limits = inferior_limit - superior_limit;
        double split = distance_between_limits / rectangles;

        while (superior_limit < inferior_limit) {
            superior_limit += split;
            sum += function(superior_limit) * split;
        }

        return -sum;
    }
}

int main() {
    double rectangles[5] = {20.0, 60.0, 1000.0, 100000.0, 1000000.0};
    double inferior_limit = -1.0;
    double superior_limit = 1.0;

    FILE *file = fopen("data.dat", "w");
    fprintf(file, "iteraciones tiempo_ms resultado\n");

    int i = 0;
    for (i = 0; i < 5; i++) {
        double initial_time = omp_get_wtime();
        double area = calculate_integral_using_riemann_sum(inferior_limit, superior_limit, rectangles[i]);
        double final_time = omp_get_wtime();
        double time_passed = (final_time - initial_time) * 1000.0;

        printf("Si hallamos el área de la función e^(-x²) en el intervalo [%.1f, %.1f] logramos: %.10f\n", inferior_limit, superior_limit, area);
        printf("Con %.1f iteraciones, el tiempo de ejecución fue de %f ms \n", rectangles[i], time_passed);
        printf("\n");

        fprintf(file, "%.0f %.6f %.10f\n", rectangles[i], time_passed, area);
    }

    fclose(file);
    return 0;
}
```

Inicialmente, incluimos como cabeceras, el Estándar de entrada y salida de información `<stdio.h>` para poder visualizar la información realizada en el cálculo. Adicionalmente, añadimos la cabecera `<math.h>` para "importar" operaciones y expresiones matemáticas importantes para esta y otras pruebas. En este caso, lo utilizamos para poder utilizar la expresión `exp()`, que representa, básicamente, el número $e$ elevado a lo que haya en el paréntesis.

Definimos una función `double function(double x) {}` para definir la función o curva la cual será nuestro objeto de cálculo. Recibe un número $x$ y devuelve la imágen que da la función al evaluarla con ese número $x$; basicamente, con esta función hallamos la "altura" del rectángulo infinitesimal.

Aquí encontramos la parte más importante para el cálculo de la integración numérica. Es un algoritmo verdaderamente simple, implementamos una función:

`double calculate_integral_using_riemann_sum(double inferior_limit, double superior_limit, double rectangles)`

En la que, inicialmente, se reciben los límites superior e inferior de la función, además de la cantidad de divisiones que se desea hacer; se define una variable crucial: `double sum` cuya función es acumular el área de cada uno de los rectángulos.

Posterior a eso, se manejan 3 casos:

1. El **límite inferior** es efectivamente menor al **límite superior**: Inicialmente, se calcula la distancia entre los límites (restando el inferior al superior) y se guarda en una variable `double distance_between_limits`, se halla la medida de la base de cada uno de los rectángulos (dividiendo la medida hallada entre el total de rectángulos), se guarda en una variable `double split` y posteriormente, se entra en un ciclo `while` que se itera hasta que el límite inferior deja de ser menor que el superior (eso significaría que el intervalo ya fue pasado); en cada iteración se le va sumando la base del rectángulo a el límite inferior, se evalúa el punto (en x, `inferior_limit`) alcanzado en la función, se multiplica por la base del rectángulo y se acumula en la variable `sum`.

2. Los límites son **iguales**: Simplemente se retorna 0, pues el área es 0.

3. EL **límite inferior** es mayor al **límite superior**: El algoritmo es el mismo que en el caso contrario, solo que ahora la variable `superior_limit` debe alcanzar al límite inferior. El área se devuelve negativa por el propiedades de la **integral definida**.



Ahora, listo el algoritmo de integración numérica, podemos proceder con la ejecución del mismo; en la salida, vemos que a más iteraciones, el valor se va acercando cada vez más a $1.4936 \ [u²]$, que, en la práctica, es el valor de el área necesitada.

En la función `main() {}` podemos utilizar las funciones planteadas inicialmente, sin embargo, para efectos prácticos y de **análisis**, realizaremos el ejercicio con cantidades diferentes de rectángulos, para ver las diferencias de convergencia y tiempo de ejecución, a medida que se va aumentando el número de **iteraciones**; para realizar este análisis de forma visual, utilizaremos GNUPlot para crear visualizaciones concretas sobre las estadísticas de las iteraciones.

Inicializamos un arreglo con diferentes cantidades de rectángulos (desde **veinte** iteraciones hasta **un millón**), utilizamos variables para almacenar los límites de **integración**, luego realizamos un ciclo `for` en el que realizamos el cálculo para cada número de rectángulos en el arreglo `rectangles[5]`, midiendo el tiempo de ejecución con la función `omp_get_wtime()`. Sin embargo, debemos tener en cuenta otra cuestión importante, y es que, a medida de que las iteraciones son mayores **¿En qué punto empieza a ser convergente al valor que queremos lograr?**, **¿Cuándo deja de ser óptimo seguir iterando para exactitud?**. Como vemos, logramos el algoritmo definitivo para la integración, pues es eficiente, rápido, llega a valores aproximados reales y computacionalmente es verdaderamente simple, pues su complejidad algorítmica es lineal ($O(n)$) y su complejidad espacial es aún menor; es, de forma clara, un algoritmo exacto, útil, sencillo y preciso... **¿o no?**.

Empecemos por algo muy importante, y es que, hay errores en la idea anterior. Riemann es un método de integración funcional, pero al evaluar su eficiencia, tiene fallos importantes:

1. El error depende **directamente** del paso, es decir al aproximar el área con escalones, el error se vuelve dependiente a la cantidad de divisiones que se hagan; es una idea simple, la premisa de la **Sumatoria de Riemann**, sin embargo, eso lo convierte en algo poco óptimo, pues al tener una dependencia tan directa al número de iteraciones, el algoritmo es, en realidad, poco eficiente, pues para obtener precisión debes someterte a un número de pasos contínuos verdaderamente grande.
2. Cada iteración **acumula** un error constante, pues depende directamente del paso, lo cual hace que incluso en iteraciones altas, halla cierto nivel de error en el cálculo.
3. Este método **converge excesivamente lento**, pues, como lo vimos, apenas llega a un resultado verdaderamente cercano luego de **cien mil** pasos, y a partir de él, la mejora de convergencia es verdaderamente lenta.

En síntesis, es un método que es **muy dependiente** a los pasos, y que necesita de **muchos**, pero **muchos** puntos para estabilizarse, a pesar de su simpleza. Además, dependiendo de como se sitúen los rectángulos debajo de la gráfica, se va a aproximar mejor o peor la curvatura, pues no es lo mismo que los rectángulos toquen a la gráfica con su vértice *superior derecho*, a que lo hagan con el *superior izquierdo*, o a que lo hagan con la *mitad de su base en el lado superior*, en todos los casos hay una diferencia radical en el cálculo.
