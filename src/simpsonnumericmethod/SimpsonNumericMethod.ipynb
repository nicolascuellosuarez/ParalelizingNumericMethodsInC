## **Las Bases del Cálculo de Áreas.**

Si estamos en un contexto universitario, el "ABC" del ingeniero corresponde a el conocimiento en los inicios paradigmáticos del cálculo, es decir, todo se construye a partir de principios tan claves como lo son los procesos de **derivación** y **antiderivación**. Formalmente, conocemos a la derivada como el cambio infinitamente inmediato que existe en una función univariable, vemos que la definición formal de la derivada corresponde a:

$$f'(x) = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}$$


Es decir, un límite de $h$ a $0$ en el que $h$ corresponde al cambio **(en $x$)** de la función; este proceso nos permite hallar la **recta tangente** a un punto específico en una función, pues la diferencia se hace infinitamente pequeña. A partir de esto, nacen múltiples métodos de derivación para hallar las derivadas de cualquier función (desde funciónes *polinómicas* a *trigonométricas*, *inversas trigonométricas*, etc.); Esto nos hace notar algo, y es que, al derivar una función, estas aplicando ciertas reglas que idealmente, pueden deshacerse, es decir, al derivar, **estás transformando una función a partir de reglas**. 

Si hacemos un ejercicio mental breve, podemos concluir que, si podemos derivar una función, podemos ***antiderivarla*** luego. Así pues, tenemos otra herramienta clave, la **antiderivada (o primitiva de una función)** como operación inversa a la **derivada**, que se define como el proceso por el que se halla una función primitiva **$F(x)$**, que al derivarla, logramos nuevamente la función original **$f(x)$** tal que:

$$F'(x) = f(x)$$

Y si **$F(x)$** es una **antiderivada** particular, podemos expresar la operación de **antiderivada** mediante la **integral indefinida** que representa una **familia de curvas**:

$$\int f(x) \ dx = F(x) + C$$

Así pues, construimos herramientas poderosas para el **análisis**, sin embargo, todo esto está basado en el uso del álgebra para la transformación de funciones definidas; no logramos una pista para lograr valores numéricos que representen áreas en intervalos específicos, hasta ahora, es mero uso del álgebra para operar funciones tanto **derivándolas** como **antiderivándolas**.

Para responder a esta necesidad, nace una herramienta mucho más poderosa aún, la **integral definida**, que nos permite, básicamente, hallar el **área** que existe debajo de la **curva o recta** de una función **$f(x)$**, en un **intervalo específico**. 

**¿Cómo se calcula esta área?**: muy simple, a partir de las herramientas construidas anteriormente: El método clásico y de perogrullo que habitualmente es enseñado, es el de, hallar la **integral indefinida** de la función cuya área se quiere encontrar, y posterior a eso, **evaluar** en unos límites de integración definidos (**$a$** y **$b$**) que corresponden a los extremos del intervalo a analizar, tal que:



$$\int_{b}^{a} f(x) \ dx$$
